
### DATASET CREATION WITH IMAGE AND TEXT (NAME & SYSTEM_ID)


```python
import os
import cv2
import pandas as pd
import tkinter as tk
from tkinter import simpledialog , messagebox
from openpyxl import load_workbook
import face_recognition
```

```python
def create_folder(name, system_id):
    folder_name = f"{system_id}_{name}"
    if os.path.exists(folder_name):
        messagebox.showwarning("Folder Exists", f"The folder '{folder_name}' already exists!")
        return None
    else:
        os.makedirs(folder_name)
        return folder_name
```

```python
def capture_images(folder_name):
    cap = cv2.VideoCapture(0)
    count = 0

    while count < 50:  # Capture 30 images
        ret, frame = cap.read()
        if ret:
            cv2.imshow('Capturing Images', frame)
            cv2.imwrite(os.path.join(folder_name, f"{count}.jpg"), frame)
            count += 1
            cv2.waitKey(100)  # Wait for 100ms between captures
        else:
            break

    cap.release()
    cv2.destroyAllWindows()
    messagebox.showinfo("Images Captured", "50 images captured successfully!")
```

```python
def update_excel(name, system_id):
    file = 'students.xlsx'
    if os.path.exists(file):
        df = pd.read_excel(file)
    else:
        df = pd.DataFrame(columns=["Name", "System ID"])

    # Check if the name and system ID already exist
    if not ((df['Name'] == name) & (df['System ID'] == system_id)).any():
        new_row = {'Name': name, 'System ID': system_id}
        df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)
        df.to_excel(file, index=False)
        print(f"Excel updated with {name} and system ID {system_id}.")
    else:
        messagebox.showwarning("Duplicate Entry", "This name and system ID already exist in the dataset.")
```

```python
def add_student():
    while True:
        name = simpledialog.askstring("Input", "Enter Student Name:")
        system_id = simpledialog.askstring("Input", "Enter System ID:")
        
        if name and system_id:
            # Update Excel with new student data
            update_excel(name, system_id)
            
            # Check for existing folder
            folder_name = f"{system_id}_{name}"
            if os.path.exists(folder_name):
                messagebox.showwarning("Folder Exists", f"The folder '{folder_name}' already exists!")
                add_more = messagebox.askyesno("Create Another?", "Do you want to create another folder for a different student?")
                if add_more:
                    continue  # Go back to the beginning of the loop
                else:
                    break
            else:
                # Create a folder for the student
                os.makedirs(folder_name)
                
                # Capture images for the student
                capture_images(folder_name)

                # Ask if user wants to add more students
                add_more = messagebox.askyesno("Add More?", "Do you want to add another student?")
                if not add_more:
                    break

    messagebox.showinfo("Done", "All student details and images have been saved successfully!")
```

```python
def run_gui():
    root = tk.Tk()
    root.withdraw()  # Hide the main window, we don't need it
    
    messagebox.showinfo("Welcome", "Let's add student details to the dataset!")
    
    add_student()
    messagebox.showinfo("Done", "All student details have been saved successfully!")

# Start the GUI
run_gui()
```

### TRANING THE MODEL AFTER DATSET IS CREATED


```python
import pandas as pd
import os
import cv2
import time
import face_recognition
import tkinter as tk
from tkinter import simpledialog, messagebox , filedialog
import pickle
```

```python
def train_classifier():
    dataset_path = 'C:/Users/Harshit/ATTANDANCE SYSTEM/AS_Dataset'  # Make sure this path is correct
    known_face_encodings = []
    known_face_names = []
    
    for folder_name in os.listdir(dataset_path):
        folder_path = os.path.join(dataset_path, folder_name)

        # Ensure we're dealing with a directory
        if os.path.isdir(folder_path):
            student_name = folder_name.split('_')[1]  # Extract the name
            
            for image_name in os.listdir(folder_path):
                image_path = os.path.join(folder_path, image_name)

                # Check if the file is an image
                if image_name.lower().endswith(('.png', '.jpg', '.jpeg')):
                    try:
                        image = face_recognition.load_image_file(image_path)
                        face_encodings = face_recognition.face_encodings(image)

                        if face_encodings:
                            known_face_encodings.append(face_encodings[0])
                            known_face_names.append(student_name)

                    except Exception as e:
                        print(f"Error loading image {image_path}: {e}")

    # Save the encodings and names
    data = {"encodings": known_face_encodings, "names": known_face_names}
    with open('face_recognition_model.pickle', 'wb') as f:
        pickle.dump(data, f)
    
    print("Classifier trained and model saved successfully!")

# Example usage
train_classifier()
```

### FACE DETECTION AND RECOGNITION


```python
import face_recognition
import cv2
import pickle
import numpy as np
import pandas as pd
```

```python
def load_model():
    with open('face_recognition_model.pickle', 'rb') as f:
        data = pickle.load(f)
    return data["encodings"], data["names"]
```

```python
def load_student_data():
    """Load student data from Excel for name and system ID matching."""
    df = pd.read_excel('students.xlsx')
    return df
```

```python
def recognize_faces():
    encodings, names = load_model()
    student_data = load_student_data()
    cap = cv2.VideoCapture(0)

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Failed to grab frame.")
            break

        rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        face_locations = face_recognition.face_locations(rgb_frame)
        face_encodings = face_recognition.face_encodings(rgb_frame, face_locations)

        for face_encoding, face_location in zip(face_encodings, face_locations):
            matches = face_recognition.compare_faces(encodings, face_encoding)
            name = "Unknown"
            system_id = "N/A"

            if True in matches:
                first_match_index = matches.index(True)
                name = names[first_match_index]
                
                # Retrieve system ID from the DataFrame
                system_id_row = student_data[student_data['Name'] == name]
                if not system_id_row.empty:
                    system_id = system_id_row.iloc[0]['System ID']

            # Draw box around face and display name and system ID
            (top, right, bottom, left) = face_location
            cv2.rectangle(frame, (left, top), (right, bottom), (0, 255, 0), 2)
            cv2.putText(frame, f"{name} (ID: {system_id})", (left, top - 10), 
                        cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)

        cv2.imshow("Face Recognition", frame)

        # Check for exit key
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()

# Example usage
recognize_faces()
```

### SECURITY AND DATA PROTECTION

```python
from cryptography.fernet import Fernet
import pickle
import os
```

```python
def save_encrypted_model(data, model_path='encrypted_model.pickle'):
    # Generate a key for encryption
    key = Fernet.generate_key()
    cipher_suite = Fernet(key)

    # Save the key securely (for this example, we save it in a file, but ideally store it safely)
    with open("encryption_key.key", "wb") as key_file:
        key_file.write(key)

    # Serialize the data and encrypt it
    serialized_data = pickle.dumps(data)
    encrypted_data = cipher_suite.encrypt(serialized_data)

    # Save encrypted model
    with open(model_path, "wb") as model_file:
        model_file.write(encrypted_data)
    print("Model encrypted and saved successfully!")
```

```python
def load_encrypted_model(model_path='encrypted_model.pickle', key_path='encryption_key.key'):
    # Load the encryption key
    with open(key_path, "rb") as key_file:
        key = key_file.read()
    cipher_suite = Fernet(key)

    # Load and decrypt the model
    with open(model_path, "rb") as model_file:
        encrypted_data = model_file.read()
    decrypted_data = cipher_suite.decrypt(encrypted_data)

    # Deserialize the data
    data = pickle.loads(decrypted_data)
    print("Model decrypted and loaded successfully!")
    return data
```
